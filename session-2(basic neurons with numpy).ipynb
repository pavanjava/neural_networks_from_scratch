### Basic Neuron


```python
x = [1, 2, 3]
weights = [0.2, 0.8, -0.5]
bias = 2

output = x[0] * weights[0] + x[1] * weights[1] + x[2] * weights[2] + bias
output
```

### Extended Basic Neuron


```python
x = [1.0, 2.0, 3.0, 2.5]
weights = [0.2, 0.8, -0.5, 1.0]
bias = 2

output = x[0] * weights[0] + x[1] * weights[1] + x[2] * weights[2] + x[3] * weights[3] + bias
output
```

### Layer of Neurons


```python
# INPUTS
x = [1, 2, 3, 2.5]

##LIST OF WEIGHTS
weights = [[0.2, 0.8, -0.5, 1],
           [0.5, -0.91, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]

##LIST OF BIASES
bias = [2, 3, 0.5]

weights1 = weights[0]
weights2 = weights[1]
weights3 = weights[2]

bias1 = bias[0]
bias2 = bias[1]
bias3 = bias[2]

outputs = [
    x[0] * weights1[0] + x[1] * weights1[1] + x[2] * weights1[2] + x[3] * weights1[3] + bias1,
    x[0] * weights2[0] + x[1] * weights2[1] + x[2] * weights2[2] + x[3] * weights2[3] + bias2,
    x[0] * weights3[0] + x[1] * weights3[1] + x[2] * weights3[2] + x[3] * weights3[3] + bias3,
]

outputs
```

### Loops for better coding the layer of neurons

```python
inputs = [1, 2, 3, 2.5]

##LIST OF WEIGHTS
weights = [[0.2, 0.8, -0.5, 1],
           [0.5, -0.91, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]

##LIST OF BIASES
biases = [2, 3, 0.5]

## FINAL OUTPUT OF EVERY NEURON
layer_outputs = []

for wt, b in zip(weights, biases):
    neuron_output = 0
    for input, input_wt in zip(inputs, wt):
        neuron_output += input * input_wt
    neuron_output += b
    layer_outputs.append(neuron_output)

layer_outputs
```

### Using Numpy for better coding

```python
import numpy as np

x = [1, 2, 3]
weights = [0.2, 0.8, -0.5]
bias = 2

output = np.dot(x, weights) + bias
output
```
