### Basic Neuron


```python
import numpy as np

x = [1, 2, 3]
weights = [0.2, 0.8, -0.5]
bias = 2

output = np.dot(x, weights) + bias
output
```

### Layer of Neurons


```python
# INPUTS
x = [1, 2, 3, 2.5]

##LIST OF WEIGHTS
weights = [[0.2, 0.8, -0.5, 1],
           [0.5, -0.91, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]

##LIST OF BIASES
bias = [2, 3, 0.5]

inputs = np.array(x)
weights = np.array(weights)
bias = np.array(bias)

output = np.dot(weights, inputs) + bias
output
```

```python
np.ndim(output)
```

```python
## Can also be done as
np.dot(inputs, weights.transpose()) + bias
```
